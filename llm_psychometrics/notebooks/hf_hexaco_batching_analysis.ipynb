```python
def read_json(file_path):
    with open(file_path, "r") as f:
        file = json.load(f)
    return file
```

```python
individual_data_dir = "hf_personas_hexaco_results_v0"
batch_data_dir = "hf_question_batching_results_v0"
```

```python
hf_individual_hexaco_results = {}
for filename in os.listdir(individual_data_dir):
    file_path = os.path.join(individual_data_dir, filename)
    file = read_json(file_path)
    hf_individual_hexaco_results =  hf_individual_hexaco_results | file
```

```python
hf_batched_hexaco_results = {}
for filename in os.listdir(batch_data_dir):
    file_path = os.path.join(batch_data_dir, filename)
    file = read_json(file_path)
    hf_batched_hexaco_results =  hf_batched_hexaco_results | file
```

```python
len(hf_individual_hexaco_results.keys()), len(hf_batched_hexaco_results.keys())
```

```python
def dict_to_df(input_dict):
    answers_df = pd.DataFrame([{"persona_id": key, "answers": input_dict[key]['answers'][0]} for key in input_dict.keys()])
    answers_df = pd.DataFrame(answers_df['answers'].to_list(), columns = list(range(0,100)), index = answers_df['persona_id'])
    
    return answers_df
```

```python
individual_hexaco_str_answers_df = dict_to_df(hf_individual_hexaco_results)
batched_hexaco_str_answers_df = dict_to_df(hf_batched_hexaco_results)
```

```python
individual_hexaco_str_answers_df.shape, batched_hexaco_str_answers_df.shape
```

```python
mapping = {"Strongly Disagree": -2, "Disagree": -1, "Neutral": 0, "Agree": 1, "Strongly Agree":2}
```

```python
individual_hexaco_answers_df = individual_hexaco_str_answers_df.replace(mapping)
batched_hexaco_answers_df = batched_hexaco_str_answers_df.replace(mapping)
```
